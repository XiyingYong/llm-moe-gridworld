```python
# ==========================================
# Cell 1: Environment Setup and Core DQN Components (Shape Fix)
# ==========================================
import numpy as np
import tensorflow as tf
from collections import deque
import matplotlib.pyplot as plt
import seaborn as sns
from maze_environment import CustomMazeEnv

import warnings
warnings.filterwarnings('ignore')
tf.get_logger().setLevel('ERROR')

# 🌟 Use the new map with greater decision-making freedom
open_map = [
    "SFFFFFG",
    "FTFPFTF",
    "FFFTFFF"
]

# Shared training hyperparameters
NUM_ITERATIONS = 800
BATCH_SIZE = 64
DISCOUNT_FACTOR = 0.95
LEARNING_RATE = 0.001

def build_dqn_model(n_states, n_actions, hidden_units=32):
    """Build a standard DQN model"""
    model = tf.keras.Sequential([
        tf.keras.Input(shape=(1,)), 
        tf.keras.layers.Embedding(input_dim=n_states, output_dim=hidden_units),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(hidden_units, activation='relu'),
        tf.keras.layers.Dense(hidden_units, activation='relu'),
        tf.keras.layers.Dense(n_actions)
    ])
    return model

def epsilon_greedy_policy(state, model, epsilon, n_actions):
    if np.random.rand() < epsilon:
        return np.random.randint(n_actions)
    else:
        # Fix 1: explicitly convert to an int32 tensor and call the model directly for inference
        state_tensor = tf.convert_to_tensor([[state]], dtype=tf.int32)
        Q_values = model(state_tensor, training=False).numpy()
        return np.argmax(Q_values[0])

def sample_experiences(replay_buffer, batch_size):
    indices = np.random.randint(len(replay_buffer), size=batch_size)
    batch = [replay_buffer[index] for index in indices]
    return [np.array([experience[field_index] for experience in batch]) for field_index in range(6)]

def training_step(batch_size, model, replay_buffer, optimizer, loss_fn, discount_factor, n_actions):
    states, actions, rewards, next_states, dones, truncateds = sample_experiences(replay_buffer, batch_size)
    
    # Fix 2: reshape to a (batch_size, 1) matrix to preserve the expected dimensions
    states = states.reshape(-1, 1)
    next_states = next_states.reshape(-1, 1)

    # Call the model directly instead of predict() for efficiency and shape safety
    next_Q_values = model(tf.convert_to_tensor(next_states, dtype=tf.int32), training=False).numpy()
    max_next_Q_values = np.max(next_Q_values, axis=1)
    runs = 1.0 - (dones | truncateds)
    target_Q_values = rewards + runs * discount_factor * max_next_Q_values
    target_Q_values = target_Q_values.reshape(-1, 1)
    mask = tf.one_hot(actions, n_actions)

    with tf.GradientTape() as tape:
        # Pass the formatted two-dimensional state matrix
        all_Q_values = model(tf.convert_to_tensor(states, dtype=tf.int32), training=True)
        Q_values = tf.reduce_sum(all_Q_values * mask, axis=1, keepdims=True)
        loss = tf.reduce_mean(loss_fn(target_Q_values, Q_values))

    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss

def train_agent(env, model, num_iterations, optimizer, loss_fn, batch_size, discount_factor):
    replay_buffer = deque(maxlen=20000)
    rewards_history = []
    
    print(f"Starting training. Total episodes: {num_iterations}")
    for episode in range(num_iterations):
        obs, _ = env.reset()
        episode_reward = 0 
        for step in range(150): 
            epsilon = max(1 - episode / (num_iterations * 0.8), 0.05) 
            
            action = epsilon_greedy_policy(obs, model, epsilon, env.action_space.n)
            next_obs, reward, done, truncated, _ = env.step(action)
            replay_buffer.append((obs, action, reward, next_obs, done, truncated))
            obs = next_obs
            episode_reward += reward

            if len(replay_buffer) > batch_size:
                training_step(batch_size, model, replay_buffer, optimizer, loss_fn, discount_factor, env.action_space.n)

            if done or truncated:
                break

        rewards_history.append(episode_reward)
        if (episode + 1) % 50 == 0:
            print(f"Episode {episode + 1}/{num_iterations} | Avg Reward(last 50): {np.mean(rewards_history[-50:]):.2f} | eps: {epsilon:.2f}")

    return model, rewards_history

def plot_training_curve(history, title, color):
    """Plot a smoothed training curve"""
    sns.set_theme(style="whitegrid")
    plt.figure(figsize=(10, 4))
    
    window = 20
    smoothed_history = np.convolve(history, np.ones(window)/window, mode='valid')
    
    plt.plot(history, alpha=0.3, color=color, label='Raw Reward')
    plt.plot(range(window-1, len(history)), smoothed_history, color=color, linewidth=2, label='Moving Average (20)')
    plt.title(f'Training Progress: {title}', fontsize=14, fontweight='bold')
    plt.xlabel('Episodes')
    plt.ylabel('Accumulated Reward')
    plt.legend()
    plt.show()

print("Environment and training components loaded. The shape issue is fixed.")
```

```python
# ==========================================
# Cell 2: Train the Goal Expert
# ==========================================
env_goal = CustomMazeEnv(expert_type='goal', custom_map=open_map)
n_states = env_goal.observation_space.n
n_actions = env_goal.action_space.n

model_goal = build_dqn_model(n_states, n_actions)
optimizer = tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE)
loss_fn = tf.keras.losses.Huber()

model_goal, history_goal = train_agent(
    env_goal, model_goal, NUM_ITERATIONS, optimizer, loss_fn, BATCH_SIZE, DISCOUNT_FACTOR
)

# Save the weights and plot the training curve
model_goal.save_weights('goal_expert_open_map.weights.h5')
plot_training_curve(history_goal, 'Goal Expert (Target: G)', '#2ca02c')
```

```python
# ==========================================
# Cell 3: Train the Prize Expert
# ==========================================
env_prize = CustomMazeEnv(expert_type='prize', custom_map=open_map)

model_prize = build_dqn_model(n_states, n_actions)
optimizer = tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE)
loss_fn = tf.keras.losses.Huber()

model_prize, history_prize = train_agent(
    env_prize, model_prize, NUM_ITERATIONS, optimizer, loss_fn, BATCH_SIZE, DISCOUNT_FACTOR
)

# Save the weights and plot the training curve
model_prize.save_weights('prize_expert_open_map.weights.h5')
plot_training_curve(history_prize, 'Prize Expert (Target: P)', '#ff7f0e')
```

```python
# ==========================================
# Cell 4: Train the Trap Expert
# ==========================================
env_trap = CustomMazeEnv(expert_type='trap', custom_map=open_map)

model_trap = build_dqn_model(n_states, n_actions)
optimizer = tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE)
loss_fn = tf.keras.losses.Huber()

model_trap, history_trap = train_agent(
    env_trap, model_trap, NUM_ITERATIONS, optimizer, loss_fn, BATCH_SIZE, DISCOUNT_FACTOR
)

# Save the weights and plot the training curve
model_trap.save_weights('trap_expert_open_map.weights.h5')
plot_training_curve(history_trap, 'Trap Expert (Target: Avoid T)', '#d62728')
```

```python
# ==========================================
# Cell 5: Train the Global Benchmark (Corrected and Reinforced Version)
# ==========================================
import gymnasium as gym

# 1. Define a wrapper that resolves hidden-state memory loss
class GlobalPOMDPWrapper(gym.Wrapper):
    """
    Expand the original 21 states to 42 states to give the agent memory:
    - States 0-20: positions before the prize is collected
    - States 21-41: positions after the prize is collected
    """
    def __init__(self, env):
        super().__init__(env)
        self.n_states_base = env.observation_space.n
        self.observation_space = gym.spaces.Discrete(self.n_states_base * 2)

    def reset(self, **kwargs):
        obs, info = self.env.reset(**kwargs)
        return obs, info

    def step(self, action):
        obs, reward, terminated, truncated, info = self.env.step(action)
        # Core logic: offset the state index after the prize is collected
        if self.env.prize_collected:
            obs += self.n_states_base
        return obs, reward, terminated, truncated, info

# 2. Create the environment and apply the wrapper
# Choose open_map or conflict_map according to the experiment; this notebook uses open_map
base_env_global = CustomMazeEnv(expert_type='global', custom_map=open_map)
env_global = GlobalPOMDPWrapper(base_env_global)

# 3. Read the state and action dimensions dynamically to avoid NameError
global_n_states = env_global.observation_space.n  # This evaluates to 42
n_actions = env_global.action_space.n

# 4. Build a wider DQN model
model_global = build_dqn_model(global_n_states, n_actions, hidden_units=64)

# 5. Apply the tuned hyperparameters from train_global_benchmark.py
# Lower the learning rate and use gradient clipping (clipnorm=1.0) for stability near traps
optimizer = tf.keras.optimizers.Adam(learning_rate=0.0005, clipnorm=1.0)
loss_fn = tf.keras.losses.Huber()

GLOBAL_ITERATIONS = 4000       # Increase the episode count to allow enough exploration
BATCH_SIZE_GLOBAL = 128        # Increase the batch size for more stable gradient updates
DISCOUNT_FACTOR_GLOBAL = 0.99  # Increase the weight of long-term returns (0.95 -> 0.99)

# 6. Start training
model_global, history_global = train_agent(
    env=env_global, 
    model=model_global, 
    num_iterations=GLOBAL_ITERATIONS, 
    optimizer=optimizer, 
    loss_fn=loss_fn, 
    batch_size=BATCH_SIZE_GLOBAL, 
    discount_factor=DISCOUNT_FACTOR_GLOBAL
)

# 7. Save the weights and plot the training curve
model_global.save_weights('global_benchmark_open_map.weights.h5')
plot_training_curve(history_global, 'Global Benchmark (Multi-Objective)', '#9467bd')
```
